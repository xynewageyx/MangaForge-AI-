# MangaForge AI — Dataset Importer
Use on Google Colab (Android browser). Only manually verified, commercially usable AI-training images may be imported.
Place images in Google Drive under `MyDrive/MangaForge_AI/source_images`. The first run creates a manifest with permissions **no** by default. Edit the CSV in Drive after reviewing licenses and run the import cell again.

```
from google.colab import drive
drive.mount('/content/drive')
from pathlib import Path
import subprocess, sys
BASE = Path('/content/drive/MyDrive/MangaForge_AI')
IMAGES = BASE / 'source_images'
IMAGES.mkdir(parents=True, exist_ok=True)
MANIFEST = BASE / 'manifest.csv'
OUTPUT = BASE / 'approved_dataset'
subprocess.run(['git', '-C', '/content/mangaforge_importer', 'fetch', 'origin', 'feature/licensed-dataset-importer', '--depth', '1'], check=True) if Path('/content/mangaforge_importer/.git').exists() else subprocess.run(['git', 'clone', '--depth', '1', '--branch', 'feature/licensed-dataset-importer', 'https://github.com/xynewageyx/MangaForge-AI-.git', '/content/mangaforge_importer'], check=True)
subprocess.run(['git', '-C', '/content/mangaforge_importer', 'checkout', '-B', 'feature/licensed-dataset-importer', 'FETCH_HEAD'], check=True) if Path('/content/mangaforge_importer/.git').exists() else None
```

```
if not MANIFEST.exists():
    subprocess.run([sys.executable, '/content/mangaforge_importer/tools/build_candidate_manifest.py', '--input', str(IMAGES), '--output', str(MANIFEST)], check=True)
    print('Review manifest.csv in Drive and set permissions to yes only when independently verified.')
else:
    print('Existing manifest preserved:', MANIFEST)
```

```
subprocess.run([sys.executable, '/content/mangaforge_importer/tools/licensed_dataset_importer.py', '--manifest', str(MANIFEST), '--output', str(OUTPUT)], check=True)
print('Results:', OUTPUT / 'accepted.csv', OUTPUT / 'rejected.csv')
```

## Find public-domain candidates (metadata only)
Search Openverse for CC0/PDM candidates. Review source pages and rights before downloading or training.

```
CANDIDATES = BASE / 'openverse_candidates.csv'
subprocess.run([sys.executable, '/content/mangaforge_importer/tools/discover_openverse.py', '--output', str(CANDIDATES), '--pages', '2'], check=True)
print('Review candidate metadata:', CANDIDATES)
```

## Audit approved images
Run only after manually reviewing and importing images. A technical pass is not a legal or semantic approval.

```
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'Pillow'], check=True)
subprocess.run([sys.executable, '/content/mangaforge_importer/tools/audit_dataset.py', '--input', str(OUTPUT / 'accepted.csv'), '--output', str(OUTPUT / 'quality_report.csv')], check=True)
print('Quality report:', OUTPUT / 'quality_report.csv')
```
